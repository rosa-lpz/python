```python
import pandas as pd
import random
import string
```

```python
# Function to generate random 3-letter combinations
def generate_random_combinations(years):
    combinations = []
    for year in years:
        for _ in range(random.randint(1, 3)):  # Random number of combinations for each year
            comb = ''.join(random.choices(string.ascii_lowercase, k=3))  # Generate 3-letter string
            combinations.append((str(year), comb))
    return combinations
```

```python
# Data generation for years 2019 to 2023
years1 = [2019, 2020, 2021, 2022, 2023]
years2 = [2019, 2021, 2022, 2023]
```

```python
# Generate 'on' and 'off' data with different random combinations
on_data = generate_random_combinations(years1)
off_data = generate_random_combinations(years1)
intersection_data = generate_random_combinations(years2)
```
