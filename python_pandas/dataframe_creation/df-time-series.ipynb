# Time series dataframe creation

```python
import pandas as pd
import numpy as np
```

## Generation of series dates and values
    * np.random.randn(100) generates 100 random numbers from a standard normal distribution.
    * .cumsum() calculates the cumulative sum to simulate a trend over time (more realistic for time series).

```python
# Creating a time series index
dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
values = np.random.randn(100).cumsum()
```

```python
dates
```

```python
values
```

## Creating a DataFrame from series data

```python
# Creating a DataFrame from series data
df = pd.DataFrame({'Date': dates, 'Value': values})
df.set_index('Date', inplace=True)
df
```
