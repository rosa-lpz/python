# Plotly - Line Charts

```python
import pandas as pd
import plotly.express as px
```

# Simple Line Chart

## From 2 lists

```python
year = [2016,2017,2018,2019,2020,2021]
count = [45,46,47,48,50,51]

# Creating the Figure instance
fig = px.line(x=year, y=count)

# showing the plot
fig.show()
```

## From a DataFrame

### Example 1

```python
df = pd.DataFrame(dict(
    x = [1, 3, 2, 4],
    y = [1, 2, 3, 4]
))
fig = px.line(df, x="x", y="y")
fig.show()
```

### Example 2

```python
# From a dataframe
# initialize list of lists (year and count)
count_year_list = [[2016,2017,2018,2019,2020,2021], [45,46,47,48,50,51]]
 
# Creating the pandas DataFrame with column the columns 'year','count'
# To keep the list of lists exactly as it is, we pass the data into the DataFrame, 
# define the index instead of the columns, and then transpose it using .T
count_by_year_df = pd.DataFrame(count_year_list, index=['year','count']).T
count_by_year_df
```

```python
# Adding x and y from dataframe
fig = px.line(x=count_by_year_df['year'], y=count_by_year_df['count'])

# showing the plot
fig.show()
```

```python
# Adding dataframe and x and y column names
fig = px.line(count_by_year_df,x='year', y='count')

# showing the plot
fig.show()
```

# Line Chart - with labes

## Example 1

```python
# Adding dataframe and x and y column names
fig = px.line(count_by_year_df,x='year', y='count', 
              title='Years vs Population', 
              labels={'year':'Year', 'count':'Population (M)'})

# showing the plot
fig.show()
```

## Example 2

```python
# Sample data
df = px.data.gapminder().query("country == 'Canada'")

# Change labels using the 'labels' argument
fig = px.line(
    df, 
    x="year", 
    y="lifeExp", 
    title='Life Expectancy in Canada',
    labels={
        "year": "Year of Observation",
        "lifeExp": "Life Expectancy (Years)"
    }
)

fig.show()
```

# Line Plots with column encoding color

```python
df = px.data.gapminder().query("continent=='Oceania'")
fig = px.line(df, x="year", y="lifeExp", color='country')
fig.show()
```

```python
df = px.data.gapminder().query("continent=='Europe'")
fig = px.line(df, x="year", y="lifeExp", color='country')
fig.show()
```

# Line charts with markers

```python
#df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country', markers=True)
fig.show()
```
