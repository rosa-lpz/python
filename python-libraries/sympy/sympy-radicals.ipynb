# Sympy - Radicals

```python
import sympy as sp
from sympy import symbols, sqrt, cbrt, root, expand, factor, simplify, radsimp, collect
```

# Overview

```python
sqrt(81)      # √x
```

```python
cbrt(216)      # ∛x
```

```python
root(32, 5)   # 5th root of x
```

## With variables (algebra)

```python
x = symbols('x')
sqrt(x)      # √x
```

```python
cbrt(x)      # ∛x
```

```python
root(x, 5)   # 5th root of x
```

# Rationalizing denominators

### $\frac{\sqrt{3}}{\sqrt{5}}$

```python
expr = sqrt(3) / sqrt(5)
# SymPy automatically rationalizes denominators
expr
```

### $\frac{\sqrt{1}}{3+\sqrt{5}}$

```python
expr = 1/(3 + sqrt(5))

# Rationalizing denominators
radsimp(expr) 
# (3 - sqrt(5))/4
```

```python
expr = sqrt(8) + sqrt(18)
expr
```

```python
simplify(expr)
# 5*sqrt(2)


radsimp(1 / (1 + sqrt(2)))
# sqrt(2) - 1
```

# Expanding and Factoring

```python
expr = (sqrt(2) + sqrt(3))**2
```

```python
expand(expr)
# 5 + 2*sqrt(6)
```

```python
expr = x*sqrt(2) + 2*sqrt(2)

factor(expr)
# sqrt(2)*(x + 2)
```

# Collecting radical terms

$3\sqrt{2} + x\sqrt{2}$

```python
expr = 3*sqrt(2) + x*sqrt(2) + 5

collect(expr, sqrt(2))
# (x + 3)*sqrt(2) + 5
```

# Solving equations with radicals

We can use `sympy` to find the value of $x$ in an equation containing radicals.

```python
from sympy import Eq, solve

x = symbols('x', real=True)

solve(Eq(sqrt(x + 1), 3), x)
# [8]
```

## Example 1 
$\sqrt(x + 5) - 3 = 0$

```python
x = sp.Symbol('x')

# Define the equation: sqrt(x + 5) - 3 = 0
equation1 = sp.Eq(sp.sqrt(x + 5) - 3, 0)
equation1
```

```python
# Solve for x
solution1 = sp.solve(equation1, x)
print("Solution to √(x + 5) - 3 = 0 is:", solution1)  # Output: [4]
```

## Example 2
$\sqrt{x} + x = 6$

```python
# Define the equation: sqrt(x) + x = 6
equation2 = sp.Eq(sp.sqrt(x) + x, 6)

# Solve for x (SymPy automatically filters out extraneous solutions)
solution2 = sp.solve(equation2, x)
print("Solution to √x + x = 6 is:", solution2)  # Output: [4]
```
