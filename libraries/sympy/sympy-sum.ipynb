# SymPy Sum

Sum a Python list with sympy.Add

```python
from sympy import Add

numbers = [1, 2, 3, 4, 5]

result = Add(*numbers)
result
```

Using Python's built-in sum (works with SymPy objects)

```python
from sympy import Integer

numbers = [Integer(1), Integer(2), Integer(3), Integer(4), Integer(5)]

result = sum(numbers)
result
```

Sum symbolic expressions

```python
from sympy import symbols#, Add

x = symbols('x')
terms = [x, 2*x, 3]
result = Add(*terms)
result
```

Symbolic summation with Sum

```python
from sympy import Sum, symbols

i = symbols('i', integer=True)

expr = Sum(i, (i, 1, 10))
print(expr)         # Unevaluated
print(expr.doit())  # Evaluated
```

**When to use each**
* sum(list) — Best for adding elements in a Python list.
* Add(*list) — Constructs a symbolic addition expression and automatically simplifies it.
* Sum(...) — Represents a mathematical summation over an index (e.g., $\sum_{i=1}^{n} i$
