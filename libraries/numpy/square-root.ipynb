# NumPy - Square root

```python
import numpy as np
```

To calculate radicals across **arrays, lists, or datasets** efficiently, `numpy` is the standard choice because it performs vectorized operations.

# Arithmetics

$\sqrt{50} = 7.07$

```python
print("Numpy square root of 50:", np.sqrt(50)) # numpy
```

$\sqrt[3]{27} = 3$ 

```python
print("NumPy cube root of 27:", np.cbrt(27)) 
```

# Arrays

```python
# Create an array of numbers
arr = np.array([4, 9, 16, 25])

# Square Root
print("Square roots:", np.sqrt(arr))   # Output: [2. 3. 4. 5.]

# Cube Root
cube_arr = np.array([8, 27, 64])
print("Cube roots:", np.cbrt(cube_arr)) # Output: [2. 3. 4.]

# Generic n-th root (e.g., 4th root)
print("Fourth roots:", np.power(arr, 1/4))
```
