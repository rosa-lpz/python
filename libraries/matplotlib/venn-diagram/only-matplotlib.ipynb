# Venn diagram using only matplotlib

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

## 2D diagram

### Example 1
* Two overlapping circes represent Set A and Set B
* Labels indicate A only, A ∩ B, and B only

```python
# Function to create a Venn diagram with two sets
def venn2():
    # Create a figure and axis
    fig, ax = plt.subplots()

    # Create two circles (for two sets) 
    # at coordinates (-0.5, 0) and (0.5, 0)
    # Transparency adjusted at 0.5 (alpha=0.5)
    circle1 = plt.Circle((-0.5, 0), 1, color='blue', alpha=0.5, label='Set A')
    circle2 = plt.Circle((0.5, 0), 1, color='red', alpha=0.5, label='Set B')

    # Add circles to the plot 
    # Also used ax.add_patch(circle1)
    ax.add_artist(circle1)
    ax.add_artist(circle2)

    # Set aspect ratio to be equal, so circles are not distorted
    ax.set_aspect('equal')

    # Set limits and remove axes for clarity
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.axis('off')  # Hide axes

    # Add labels to the intersection and sets
    ax.text(-0.5, 0.5, 'A', fontsize=20, color='blue', ha='center')
    ax.text(0.5, 0.5, 'B', fontsize=20, color='red', ha='center')
    ax.text(0, 0, 'A ∩ B', fontsize=12, color='black', ha='center')

    plt.title('Simple Venn Diagram (2 sets)')
    plt.show()

# Call the function to display the Venn diagram
venn2()
```

### Example 2
* Two overlapping circes represent Set A and Set B (list of strings)
* Labels indicate A only, A ∩ B, and B only

```python
# Create two datasets -----
# Set A: Products purchased by Customer 1
data1 = ['Laptop', 'Smartphone', 'Headphones', 'Smartwatch', 'Tablet']
set_a = pd.Series(data1)

# Set B: Products purchased by Customer 2
data2 = ['Smartphone', 'Smartwatch', 'Mouse', 'Keyboard', 'Laptop']
set_b = pd.Series(data2)

# Calculate the intersection and unique elements -----
intersection = pd.Series(np.intersect1d(set_a, set_b))  # Common products
unique_a = pd.Series(set_a[~set_a.isin(set_b)])  # Products unique to Set A (not in Set B)
unique_b = pd.Series(set_b[~set_b.isin(set_a)])  # Products unique to Set B (not in Set A)
```

```python
unique_b
```

```python
# Venn diagram using only Matplotlib -----

# Create a figure and axis
fig, ax = plt.subplots()

# Create two circles (for two sets)
circle1 = plt.Circle((-0.5, 0), 1, color='blue', alpha=0.5, label='Set A (Customer 1)')
circle2 = plt.Circle((0.5, 0), 1, color='green', alpha=0.5, label='Set B (Customer 2)')

# Add circles to the plot
ax.add_artist(circle1)
ax.add_artist(circle2)

# Set aspect ratio to be equal, so circles are not distorted
ax.set_aspect('equal')

# Set limits and remove axes for clarity
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.axis('off')  # Hide axes

# Add text to represent the sets and their intersection
ax.text(-0.5, 0.5, 'Set A', fontsize=12, color='white', ha='center')
ax.text(0.5, 0.5, 'Set B', fontsize=12, color='white', ha='center')

# Add the intersection and unique elements text
# ax.text(x, y, text): This function adds text to the plot at the position (x, y) on the plot.
# f'Intersection: {", ".join(intersection)}'This is an f-string (formatted string) that dynamically inserts the value of intersection into the string.
ax.text(0, 0, f'Intersection: {", ".join(intersection)}', fontsize=10, color='black', ha='center')
ax.text(-0.8, -0.4, f'Unique to A: {", ".join(unique_a)}', fontsize=10, color='black', ha='center')
ax.text(0.8, -0.4, f'Unique to B: {", ".join(unique_b)}', fontsize=10, color='black', ha='center')

# Title of the Venn diagram
plt.title('Venn Diagram of Products Purchased by Two Customers')

# Show the plot
plt.show()
```

```python
# Venn diagram using only Matplotlib -----

# Create a figure and axis
fig, ax = plt.subplots()

# Create two circles (for two sets)
circle1 = plt.Circle((-0.5, 0), 1, color='blue', alpha=0.5, label='Set A (Customer 1)')
circle2 = plt.Circle((0.5, 0), 1, color='green', alpha=0.5, label='Set B (Customer 2)')

# Add circles to the plot
ax.add_artist(circle1)
ax.add_artist(circle2)

# Set aspect ratio to be equal, so circles are not distorted
ax.set_aspect('equal')

# Set limits and remove axes for clarity
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.axis('off')  # Hide axes

# Add text to represent the sets and their intersection
ax.text(-0.7, 0.5, 'Set A', fontsize=12, color='white', ha='center')
ax.text(0.7, 0.5, 'Set B', fontsize=12, color='white', ha='center')

# Add the intersection and unique elements text
ax.text(0, 0, f"{intersection.count()}", fontsize=10, color='black', ha='center')
ax.text(-0.9, -0.4, f'Unique to A: {unique_a.count()}', fontsize=10, color='black', ha='center')
ax.text(0.9, -0.4, f'Unique to B: {unique_b.count()}', fontsize=10, color='black', ha='center')

# Title of the Venn diagram
plt.title('Venn Diagram of Products Purchased by Two Customers')

# Show the plot
plt.show()
```
