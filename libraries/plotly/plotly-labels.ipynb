# Labels in plotly

```python
import pandas as pd
import plotly.express as px
```

## Using the labels argument in Plotly Express
When using Plotly Express (px.line), we can pass a dictionary to the labels parameter. The keys should be your original column or data names, and the values should be the custom labels you want to display.

```python
# Sample data
df = px.data.gapminder().query("country == 'Canada'")

# Change labels using the 'labels' argument
fig = px.line(
    df, 
    x="year", 
    y="lifeExp", 
    title='Life Expectancy in Canada',
    labels={
        "year": "Year of Observation",
        "lifeExp": "Life Expectancy (Years)"
    }
)

fig.show()
```

## Using update_layout
If you we already generated the figure object (using either Plotly Express or Plotly Graph Objects), we can change the axis titles by passing xaxis_title and yaxis_title to .update_layout().

```python
fig = px.line(x=[1, 2, 3], y=[10, 20, 30])

# Update axis titles
fig.update_layout(
    xaxis_title="Custom X-Axis Label",
    yaxis_title="Custom Y-Axis Label"
)

fig.show()
```

## Using update_xaxes and update_yaxes
Alternatively, we can target the axes individually using .update_xaxes() and .update_yaxes() with the title_text parameter. This is especially useful if you are working with multi-axis charts or subplots.

```python
fig = px.line(x=[1, 2, 3], y=[10, 20, 30])

# Update individual axes
fig.update_xaxes(title_text="Custom X-Axis Label")
fig.update_yaxes(title_text="Custom Y-Axis Label")

# You can also style the label font size/color here:
# fig.update_xaxes(title_text="Custom X Label", title_font=dict(size=18, color='red'))

fig.show()
```
