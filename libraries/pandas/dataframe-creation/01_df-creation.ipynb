# Dataframe Creation - Quick guide


## Pandas Create Dataframe Syntax

pandas.DataFrame(data, index, columns) 

**Parameters:** 
    * data: It is a dataset from which a DataFrame is to be created. It can be a list, dictionary, scalar value, series, and arrays, etc. 
    * index: It is optional, by default the index of the DataFrame starts from 0 and ends at the last data value(n-1). It defines the row label explicitly. 
    * columns: This parameter is used to provide column names in the DataFrame. If the column name is not defined by default, it will take a value from 0 to n-1. 

**Returns:**
* DataFrame object

#### Example 3
If we want to keep the list of lists exactly as it is, we can pass the data into the DataFrame, define the index instead of the columns, and then transpose it using .T

```python
count_year_list = [[2016, 2017, 2018, 2019, 2020, 2021], [45, 46, 47, 48, 50, 51]]

# Define them as rows first, then flip (transpose) them into columns
count_by_year_df = pd.DataFrame(count_year_list, index=['year', 'count']).T

count_by_year_df
```

### From List of Tuples

```python
# initialize list of tuples
data = [
    (1, 'Alice', 25),
    (2, 'Bob', 30),
    (3, 'Charlie', 35)
]

# Creating DataFrame with columns 'ID', 'Name', 'Age'
df = pd.DataFrame(data, columns=['ID', 'Name', 'Age'])

# Displaying DataFrame
df.head()
```

### From List of Dictionaries


```python
# Initialize data to lists.
data = [{'a': 1, 'b': 2, 'c': 3},
        {'a': 10, 'b': 20, 'c': 30}]

# Creating DataFrame, keys as columns
df = pd.DataFrame(data)

# Displaying DataFrame
df.head()
```

#### Explanation
* 'a', 'b', and 'c' are the column names.
* The values inside the dictionaries (e.g., 1, 2, 3 in the first dictionary, 10, 20, 30 in the second) represent the actual values for each column in that row.

## Creating DataFrame from Dictionaries
### From a Dictionary of narray/lists

To create DataFrame from dict of narray/list, all the narray must be of same length. If index is passed then the length index should be equal to the length of arrays. If no index is passed, then by default, index will be range(n) where n is the array length.

#### Example 1

```python

# DataFrame from dict narray / lists

 
# initialize data of lists.
data_m3 = {'Name': ['Tom', 'nick', 'krish', 'jack'],
        'Age': [20, 21, 19, 18]}
 
# Create DataFrame
df_m3 = pd.DataFrame(data_m3)
 
# Print the output.
df_m3
```

#### Example 2

```python
# Directly map column names to their respective lists
count_by_year_df = pd.DataFrame({
    'year': [2016, 2017, 2018, 2019, 2020, 2021],
    'count': [45, 46, 47, 48, 50, 51]
})

count_by_year_df
```

## Creating DataFrame using Zip
Generate first list of lists is being and then we can pair the elements up row-by-row using zip().

### From a List of List/Arrays and then Zip()

```python
count_year_list = [[2016, 2017, 2018, 2019, 2020, 2021], [45, 46, 47, 48, 50, 51]]

# zip(*list) unpacks the two sublists and pairs them up: (2016, 45), (2017, 46), etc.
count_by_year_df = pd.DataFrame(list(zip(*count_year_list)), columns=['year', 'count'])

count_by_year_df
```
