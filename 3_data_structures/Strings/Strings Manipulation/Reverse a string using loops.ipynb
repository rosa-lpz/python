# Reverse a string using loops

## Example 1

```python
def reverse(s):
	str = ""
	for i in s:
		str = i + str
	return str

s = "Geeksforgeeks"

print("The original string is : ", end="")
print(s)

print("The reversed string(using loops) is : ", end="")
print(reverse(s))
```

### Example 1- Code to see each iteration

```python

def reverse(s):
	str = ""
	for i in s:
		str = i + str
		print(str)
	return str

s = "Geeksforgeeks"

print("The original string is : ", end="")
print(s)

print("The reversed string(using loops) is : ", end="")
print(reverse(s))
```
