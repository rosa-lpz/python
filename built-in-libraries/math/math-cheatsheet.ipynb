# Math Built-in Library

```python
import math
```


The standard `math` library is best for quick, **numerical evaluations** of real numbers. It provides a direct function for square roots and can handle any $n$-th root using exponentiation.

# Square Root

$\sqrt{50}$

```python
sqrt_val = math.sqrt(50)
print(f"Square root of 50: {sqrt_val}")  # Output: 7.0710678118654755
```

$\sqrt[3]{27} = 3$ 

```python
cube_root = math.pow(27, 1/3)
print(f"Cube root of 27: {cube_root}")    # Output: 3.0
```

$\sqrt[4]{256} = 4.0$ 

```python
# Generic n-th Root: ⁴√256
fourth_root = math.pow(256, 1/4)
print(f"Fourth root of 256: {fourth_root}") # Output: 4.0
```
