# Matplotlib Venn diagram


```python
import pandas as pd
from matplotlib import pyplot as plt
from matplotlib_venn import venn2 # use python3 -m pip install matplotlib-venn to install matplotlib_venn in environment
```

## 2D diagram

```python
# Create two sets
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Create Venn diagram
plt.figure(figsize=(6,6))
venn = venn2([set_a, set_b], set_labels=('Set A', 'Set B'))

# Show the plot
plt.title("Simple Venn Diagram")
plt.show()
```
