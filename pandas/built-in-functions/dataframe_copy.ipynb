# DataFrame.copy(deep=True)
In **pandas**, the `DataFrame.copy()` method creates a deep or shallow copy of a DataFrame. It’s useful when you want to manipulate or modify a DataFrame without affecting the original DataFrame. The default behavior is to make a deep copy, but you can specify a shallow copy if needed.

## Syntax:

```python
DataFrame.copy(deep=True)
```

- **deep** (default `True`): If `True`, it makes a deep copy. This means that it copies both the DataFrame's structure and its data, so changes to the new DataFrame won't affect the original.
    
- **deep=False**: If `False`, it creates a shallow copy. The new DataFrame will share the underlying data with the original DataFrame, meaning changes to the data in the new DataFrame will also reflect in the original one.

## Examples:

### Deep Copy (default):

````python
import pandas as pd

# Original DataFrame
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

# Make a deep copy
df_copy = df.copy()

# Modify the copy
df_copy['A'] = [7, 8, 9]

# Print both DataFrames
print("Original DataFrame:")
print(df)
print("\nCopied DataFrame:")
print(df_copy)
````

### Shallow Copy:

````python
# Make a shallow copy
df_shallow_copy = df.copy(deep=False)

# Modify the shallow copy
df_shallow_copy['A'] = [10, 11, 12]

# Print both DataFrames
print("Original DataFrame:")
print(df)
print("\nShallow Copied DataFrame:")
print(df_shallow_copy)
````

In the case of a shallow copy, changes to the `df_shallow_copy` will affect the original DataFrame because the data is shared.

#### When to use Deep vs. Shallow copy:

- **Deep copy** is useful when you want to make sure changes to the copy don’t impact the original data.
    
- **Shallow copy** can be useful for performance optimization when you know you don't need to modify the underlying data.
